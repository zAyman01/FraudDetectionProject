```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X_train = pd.read_csv("../processed/X_train.csv")
y_train = pd.read_csv("../processed/y_train.csv").squeeze()
```

```python
missing = X_train.isnull().sum()
print("Missing Values (Top 10):\n", missing[missing > 0].sort_values(ascending=False).head(10))
```

```python
print("\nTarget Distribution:\n", y_train.value_counts(normalize=True))

y_train.value_counts().plot(kind='bar')
plt.title("Target Distribution")
plt.show()
```

```python
X_sample = X_train.sample(n=5000, random_state=42)
y_sample = y_train.loc[X_sample.index]
```

```python
num_cols = X_sample.select_dtypes(include=np.number).columns[:10]

X_sample[num_cols].hist(figsize=(12,8))
plt.suptitle("Distributions After Preprocessing")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(X_sample[num_cols].corr(), cmap='coolwarm')
plt.title("Correlation Check")
plt.show()
```

```python
sns.boxplot(data=X_sample[num_cols])
plt.xticks(rotation=90)
plt.title("Outliers After Preprocessing")
plt.show()
```

```python
print("\nFinal Checks Done:")
print("- Missing values handled")
print("- Target imbalance verified")
print("- Feature distributions checked")
print("- Correlation reviewed")
print("- Outliers inspected")
```
